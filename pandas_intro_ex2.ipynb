## Passo 0 — Importar pandas e carregar os dados

```python
import pandas as pd
df = pd.read_parquet("/content/emendas_parlamentares (1).parquet")

```

## 1) Filtre apenas as emendas do ano de **2022** e mostre as 30 primeiras linhas. Em seguida responda: Quanta emendas foram registradas em 2022?

```python
df_2022 = df[df['Ano da Emenda'] == 2022]
print("As 30 primeiras emendas de 2022:")
print(df_2022.head(30))
total_emendas_2022 = len(df_2022)
print(f"\nForam registradas {total_emendas_2022} emendas em 2022.")
```

## 2) Filtre as emendas da região **Norte** que tenham `Valor Empenhado` maior que **500.000** e mostre as 5 primeiras linhas. Em seguida, responda: quantas emendas atendem a esses duas condições ao mesmo tempo?

```python
df_norte_500k = df[(df['Região'] == 'NORTE') & (df['Valor Empenhado'] > 500000)]
print("--- Primeiras 5 emendas da Região Norte > 500k ---")
display(df_norte_500k.head(5))
total_condicoes = len(df_norte_500k)
print(f"\nExistem {total_condicoes} emendas que são da região Norte e possuem valor empenhado superior a R$ 500.000,00.")
```

## 3) Filtre as emendas que são do estado de **SÃO PAULO** ou do estado do **RIO DE JANEIRO** e mostre quantas linhas cada estado possui. Em seguida, responda: Qual dos dois estados possui mais emendas e quantas são?

```python
df_rj = df[df['Município'] == 'RIO DE JANEIRO']
df_sp = df[df['Município'] == 'SÃO PAULO']
sp_count = len(df_sp)
rj_count = len(df_rj)
print(f"O estado de São Paulo possui {sp_count} emendas.")
print(f"O estado do Rio de Janeiro possui {rj_count} emendas.")

```

## 4) Na coluna `Código da Emenda`, substitua todos os valores `"Sem informação"` por `"Não disponível"`. Mostre a contagem de valores únicos da coluna antes e depois da substituição.

```python
df_sem_info = df[df['Código da Emenda'] == 'Sem informação']
print("Valores únicos antes da substituição:")
print(df['Código da Emenda'].unique())
df['Código da Emenda'] = df['Código da Emenda'].replace('Sem informação', 'Não disponível')
print("\nValores únicos após a substituição:")
print(df['Código da Emenda'].unique())
```

## 5) Na coluna `Nome Função`, transforme todos os valores para que fiquem em letras maiúsculas. Confirme a alteração mostrando os valores únicos da coluna.

```python
df = df.copy()
df['Nome Função'] = df['Nome Função'].str.upper()
print(df['Nome Função'].unique())
```

## 6) Verifique quantas linhas duplicadas existem no dataset. Em seguida, remova as duplicatas e mostre quantas linhas sobraram.

```python
df_duplicados = df[df.duplicated()]
print(f"Existem {len(df_duplicados)} linhas duplicadas no DataFrame.")
df = df.drop_duplicates()
print(f"Após a remoção de duplicatas, restam {len(df)} linhas no DataFrame.")
```

## 7) Remova duplicatas considerando apenas as colunas `Código da Emenda` e `Município`. Quantas linhas sobraram? O resultado é diferente de quando removemos duplicatas do DataFrame inteiro?

```python
df = df.drop_duplicates(subset=['Código da Emenda', 'Município'])
print(f"Após a remoção de duplicatas, restam {len(df)} linhas no DataFrame.")
df_duplicados = df[df.duplicated(subset=['Código da Emenda', 'Município'])]
print(f"Existem {len(df_duplicados)} linhas duplicadas no DataFrame considerando as colunas 'Código da Emenda' e 'Município'.")
```

## 8) Identifique quais colunas possuem valores nulos e quantos nulos cada uma tem. Em seguida, responda: qual é a coluna que possui maior número de valores nulos e quantos são?

```python
df_nulos = df.isnull().sum()
print("Número de valores nulos em cada coluna:")
print(df_nulos)
coluna_maior_nulos = df_nulos.idxmax()
print(f"\nA coluna com maior número de valores nulos é '{coluna_maior_nulos}' com {df_nulos[coluna_maior_nulos]} valores nulos.")
```

## 9) Preencha os valores nulos da coluna `codigo do autor` com **-1** e os valores nulos da coluna `nome do autor da emenda` com "desconhecido" e confirme que não restaram nulos nessa coluna.

```python
df.columns
```

```python
df = df.copy()
df['Código do Autor da Emenda'] = df['Código do Autor da Emenda'].fillna(-1)
df['Nome do Autor da Emenda'] = df['Nome do Autor da Emenda'].fillna('desconhecido')
print("Valores nulos após a substituição:")
print(df.isnull().sum())
```

## 10) Verifique o tipo da coluna `Ano da Emenda`. Converta-a para **string** (`str`). Mostre o tipo antes e depois da conversão.

```python
df = df.copy()
print("Tipo da coluna 'Ano da Emenda' antes da conversão:")
print(df['Ano da Emenda'].dtype)
df['Ano da Emenda'] = df['Ano da Emenda'].astype(str)
print("\nTipo da coluna 'Ano da Emenda' após a conversão:")
print(df['Ano da Emenda'].dtype)
```

## 11) Converta a coluna `Valor Empenhado` para o tipo **float** (caso ainda não seja). Mostre o tipo após a conversão.

```python
print("Tipo da coluna 'Valor Empenhado' antes da conversão:")
print(df['Valor Empenhado'].dtype)
df['Valor Empenhado'] = df['Valor Empenhado'].astype(float)
print("\nTipo da coluna 'Valor Empenhado' após a conversão:")
print(df['Valor Empenhado'].dtype)
```

## 12) Calcule o **total** de `Valor Pago` por **Região**. Quais são as duas regiões que mais receberam recursos?

```python
total_pago_por_regiao = df.groupby('Região')['Valor Pago'].sum().sort_values(ascending=False)
print("Total de Valor Pago por Região:")
print(total_pago_por_regiao)

primeiras_duas_regioes = total_pago_por_regiao.head(2)
print("\nAs duas regiões que mais receberam recursos são:")
print(primeiras_duas_regioes)
```

## 13) Calcule a **quantidade** de emendas por **Nome Função**. Mostre as 5 funções com mais emendas.

```python
emendas_por_funcao = df.groupby('Nome Função').size().sort_values(ascending=False)
print("Quantidade de emendas por Nome Função:")
print(emendas_por_funcao.head(5))
```

## 14) Crie uma nova coluna chamada `Saldo a Pagar`. Ela deve ser o resultado da subtração entre a coluna `Valor Empenhado` e a coluna `Valor Pago`. Mostre as 5 primeiras linhas apenas com essas três colunas.

```python
df = df.copy()
df['Saldo a Pagar'] = df['Valor Empenhado'] - df['Valor Pago']
print(df[['Valor Empenhado', 'Valor Pago', 'Saldo a Pagar']].head())
print(df.columns)
print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

## 15) Elimine a(s) coluna(s) do dataframe que eestá totalmente vazia.

```python
print(f"Shape do DataFrame antes da remoção de colunas vazias: {df.shape}")
df_columns_before = df.columns

df = df.dropna(axis=1, how='all')

columns_removed = [col for col in df_columns_before if col not in df.columns]

if columns_removed:
    print(f"\nColuna(s) removida(s): {', '.join(columns_removed)}")
else:
    print("\nNenhuma coluna totalmente vazia foi encontrada para remoção.")

print(f"Shape do DataFrame após a remoção de colunas vazias: {df.shape}")
```

## 16) Para finalizar, salve o DataFrame transformado em um arquivo no formato JSON.

```python
df.to_json('emendas_transformado.json', orient='records', indent=4)
print("DataFrame salvo como 'emendas_transformado.json'!")
```
